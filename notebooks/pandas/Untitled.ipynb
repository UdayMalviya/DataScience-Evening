```python
s = 'Hello,Wor"l"d.'
```

```python
def clean_sp_chr(s: str) -> str:
    res = ''
    for i in s:
        if i in (',.()""'):
            continue
        # res += "".join(i)
        res = res + "".join(i)
    return res
```

```python
clean_sp_chr(s)
```
