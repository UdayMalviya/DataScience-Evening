```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    "Name": ['A', 'B', 'C', 'A', 'B', 'C'],
    'Age': [20, 22, 40, 50, 36, 20]
})
```

Senior > 40 |
40 <= mid < 30|
30 <= junior|

```python
df  # "Seniority" senior > 40, mid> 30, junior< 30
```

```python
df["Seniority"] = df['Age'].apply(lambda x: 'senior'  if x > 40 else 'mid'  if x > 30 else 'junior')
```

```python
df
```

```python
# df.assign(exp= lambda x: 'high' if x['Seniority'] == 'senior' else 'low')
def check_exp(x):
    if x == 'senior':
        return 'high'
    else:
        return 'low'
```

```python
df['Seniority'].apply(check_exp)
```
