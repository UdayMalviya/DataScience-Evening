```python
import numpy as np
from sklearn.datasets import make_regression
```

```python
X, y = make_regression(n_features=100, n_samples=1000, noise=20)
# l1- LASSO 
# L2- Ridge 
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.2)
```

```python
from sklearn.linear_model import Lasso, Ridge, LinearRegression
```

```python
model = Lasso(alpha=0.001)
```

```python
model.fit(X_train, y_train)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
reg = LinearRegression()
reg.fit(X_train, y_train)
reg.coef_, reg.intercept_
```

```python
from sklearn.metrics import  root_mean_squared_error
```

```python
root_mean_squared_error(y_test, model.predict(X_test))
```

```python
root_mean_squared_error(y_test, reg.predict(X_test))
```

```python

d = {}
for i in range(0, 100): # loop 100 times
    temp = {} # temp dict to hold coef, intercept
    n  = np.random.rand() # generate a random num (0-1)
    model = Lasso(alpha=n) 
    model.fit(X_train, y_train) # fitting model
    # coef = model.coef_ # calculation coef_
    # intercept = model.intercept_ # calculation intercept_
    # temp['coef_'] = coef  # adding coef to temp
    # temp['intercept_'] = intercept # adding intercept to temp
    y_pred = model.predict(X_test)
    rmse = root_mean_squared_error(y_test, y_pred)
    temp['rmse'] = rmse
    temp['alpha'] = model.alpha
    d[i] = temp 
    # print(n, end=' ')
    
    
# {0: {coef_: , intercept_:}}
```
