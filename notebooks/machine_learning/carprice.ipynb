```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import r2_score, mean_squared_error
```

```python
df = pd.read_csv('../../Data/CarPrice_Assignment.csv')
```

```python
df
```

```python
X = df.drop(columns=['CarName', 'price'])
y = df['price']
```

```python
X = X.sample(len(df))
```

```python
number = X.select_dtypes('number')
categories = X.select_dtypes('string')
```

```python
number.shape, categories.shape
```

```python
X.shape
```

```python
tranformer = ColumnTransformer([
    ('scaling', StandardScaler(), number.columns),
    ('ohe', OneHotEncoder(drop='first', sparse_output=False), categories.columns)
], remainder='passthrough')

transformer2 = ColumnTransformer([
    ('ohe', OneHotEncoder(drop='first', sparse_output=False), categories.columns)
], remainder='passthrough')
```

```python
linear_reg_pipe = Pipeline([
    ('transform', tranformer),
    ('model', LinearRegression())
])
```

```python
decision_tree_pipe = Pipeline([
    ('transform', transformer2 ),
    ('model', DecisionTreeRegressor())
])
```

```python
models = {'linear': linear_reg_pipe, 'decision': decision_tree_pipe}
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
for name, pipeline in models.items():
    pipeline.fit(X_train, y_train)
    print(X_test.info())
    y_pred = pipeline.predict(X_test)
    mse = mean_squared_error(y_test, y_pred)
    r2 = r2_score(y_test, y_pred)
    print(f"For {name} results are:\nMSE: {mse} and R2: {r2}")
    
```
