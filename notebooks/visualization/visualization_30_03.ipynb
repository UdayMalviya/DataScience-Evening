```python
import numpy as np
data = pd.DataFrame({"Date":pd.date_range('2024-01-01', '2024-01-31', freq='D'), 
             "numbers": np.random.rand((31))})
```

```python
data
```

```python
sns.lineplot(data, x='Date', y='numbers', estimator='median', markers=True)
```

```python
sns.scatterplot(data, x='Date', y='numbers')
```

```python
fig, ax = plt.subplot_mosaic([[1,1], [2,2]])
sns.lineplot(data, x='Date', y='numbers', ax=ax[1])
sns.lineplot(data, x='Date', y='numbers', ax=ax[2], estimator='median')
```
