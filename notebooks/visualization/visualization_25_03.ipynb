```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# sin
 
x = np.linspace(0, 4* np.pi, 1000) 
y = np.sin(x)

fig, ax = plt.subplots( figsize=(12,4))
ax.set_title('sin wave')
ax.set_xlabel("X axis")
ax.set_ylabel("Y axis")
ax.plot(x, y, '--r') # lineplot 
# ax[1].plot(x, np.cos(y), '.g',)

plt.show()
```
