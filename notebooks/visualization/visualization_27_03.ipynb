```python
arr = np.random.rand(100)
arr2 = np.random.randn(100)
```

```python
plt.text()
```

```python
fig, ax = plt.subplots(2)
ax[0].hist(arr, bins=20, color='red', alpha=0.6)
ax[0].axvline(arr.mean(), color='green', linestyle=':')
ax[0].axvline(np.median(arr), color='black', linestyle='--')
ax[0].axvline(np.std(arr), color='#052452')
ax[0].text(0.6, 6, "TEXT", rotation='horizontal', fontsize=20)
ax[1].hist(arr2, bins=20, color='red', alpha=0.6)
ax[1].axvline(arr2.mean(), color='green', linestyle=':')
ax[1].axvline(np.median(arr2), color='black', linestyle='--')
ax[1].axvline(np.std(arr2), color='#052452')
plt.show()
```
